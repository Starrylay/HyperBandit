```python
import matplotlib.pyplot as plt
import numpy as np
import pickle
```

```python

with open('./result_log/resultdata.pkl', 'rb') as f:
    algresult = pickle.load(f)
for alg_name, alg in algresult.algorithms.items():
    print(alg_name)
```

```python
for alg_name, alg in algresult.algorithms.items():
    if alg_name == 'LinUCB' or alg_name == 'DLinUCB' or alg_name == 'ColinUCB':
        plt.plot(range(len(algresult.AlgRewardRatio_vsRandom[alg_name])), algresult.AlgRewardRatio_vsRandom[alg_name], '-o', markersize=4, label=alg_name)
    else:
        plt.plot(range(len(algresult.AlgRewardRatio_vsRandom[alg_name])), algresult.AlgRewardRatio_vsRandom[alg_name], '-^', markersize=4, label=alg_name)

plt.legend()

```
