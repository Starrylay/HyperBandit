### 生成聚类向量

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
# 假设 user_vectors 是用户向量的集合，每个向量形状为 (n_features,)
Userfeatures_noid = data_feature
n_clusters = 40# 指定聚类的数量
kmeans = KMeans(n_clusters=n_clusters)
kmeans.fit(Userfeatures_noid)
centers = kmeans.cluster_centers_
for i in range(len(centers)):
    print("Class {}: {}".format(i+1, centers[i]))


# 计算轮廓系数
silhouette_avg = silhouette_score(Userfeatures_noid, kmeans.labels_)
print(f"轮廓系数：{silhouette_avg}")
```

### 生成用户聚类结果信息

```python
silhouette_avg = silhouette_score(Userfeatures_noid, kmeans.labels_)
print(f"轮廓系数：{silhouette_avg}")
print("label",  kmeans.labels_)
print("num_label",  len(kmeans.labels_))
print(kmeans.labels_[1])# test
```

### 同步存储聚类后的用户向量 和 映射关系

```python
# 因为聚类了所以不用考虑原来用户id，类别顺序就是新的id
filename = "../kuairec/{}clusters_FeatureVectors.dat".format(n_clusters)
with open(filename, 'w', newline='') as f:
    for i in range(centers.shape[0]):
        new_line = str(i) + '\t' + ';'.join(str(centers[i, :]).strip('[').strip(']').split()) + '\n'
        f.write(new_line)
print("done")

#label需要类别顺序与用户id对应，而不是与用户向量的index对应
label_dict = {}
for i in range(len(kmeans.labels_)):
    label_dict[str(data_user[i])] = int(kmeans.labels_[i])

print(label_dict)
import json
filename = "../kuairec/{}clusters_labels.json".format(n_clusters)
with open(filename, 'w') as f:
    json.dump(label_dict, f)
print("done")
```

## 生成关系矩阵

### 读取聚类向量

```python
import numpy as np
import pandas as pd
```

```python
classfeatures = []
n_clusters=40
filename = "../kuairec/{}clusters_FeatureVectors.dat".format(n_clusters)
with open(filename, "r") as f:
    # f.readline()# 跳过第一行
    for line in f:
        line = line.split("\t")
        vec = line[1].strip("[]").strip("\n").split(";") #[0] 是item id [1]是arm feature
        classfeatures.append(np.array(vec).astype(float))
classfeatures = pd.DataFrame(classfeatures)
classfeatures
       
```

### 生成W

```python
import numpy as np
import json
classfeatures = np.array(classfeatures)
n = len(classfeatures)
sparsityLevel = int(n/2)
W = np.zeros(shape=(n, n))

for i in range(n):
    sSim = 0
    for j in range(n):
        sim = np.dot(classfeatures[i], classfeatures[j])
        W[i][j] = sim
        sSim += sim
    W[i] /= sSim
SparseW = W

#小于阈值的置为0
if sparsityLevel > 0 and sparsityLevel < n:
    for i in range(n):
        similarity = sorted(W[i], reverse=False)
        threshold = similarity[sparsityLevel]
        for j in range(n):
            if W[i][j] <= threshold:
                SparseW[i][j] = 0
        SparseW[i] /= sum(SparseW[i])
# for i in range(len(SparseW)):
    
#     print(sum(SparseW[i]))

for i in range(n):
    # SparseW[i][i] = 0
    # if sum(SparseW[i]) != 0:
    #     SparseW[i][i] = np.linalg.norm(SparseW[i]) ** 2 / sum(SparseW[i])
    # else:
    SparseW[i][i] = 1
    SparseW[i] /= sum(SparseW[i])
print("SparseW --Opt", SparseW)

# for i in range(len(SparseW)):
    
#     print(sum(SparseW[i]))
#存储为json格式
# import json
filename = "../kuairec/SparseW.json"
with open(filename, 'w') as f:
    json.dump(SparseW.T.tolist(), f)
print("save SparseW.json")
# W = np.load('./NYC/SparseW.npy')
```
