```python
import torchtext
import re
import numpy as np
import nltk
nltk.download('punkt')
from nltk.tokenize import word_tokenize
import pandas as pd

categories_week = np.array(['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
categories_day = np.array(['morning', 'noon', 'afternoon', 'night', 'rest'])

embedding = torchtext.vocab.GloVe(name='840B', dim=300)
category_embedding = [] #np.zeros(shape=(35, 300), dtype=np.float32)
keys = []
for i, text1 in enumerate(categories_week):
    for j, text2 in enumerate(categories_day):
        text = text1 + ' ' + text2
        clean_text = re.sub(r'[^\w\s]','',text)
        token = word_tokenize(clean_text)
        print(token)
        key = text1[:3] + ' ' + text2
        keys.append(key)
        category_embedding.append(embedding.get_vecs_by_tokens(token).numpy().mean(axis=0))
category_df = pd.DataFrame(category_embedding)
category_df.index = keys
category_df
```

```python
from sklearn.decomposition import PCA
n_components  = 30
pca = PCA(n_components=n_components)
item_df_for_pca = category_df
pca.fit(item_df_for_pca)
item_pca = pca.transform(item_df_for_pca)
item_pca
```

```python
item_pca_df = pd.DataFrame(item_pca)
item_pca_df.index = keys
index_item_pca_df = item_pca_df.reset_index()#.rename(columns={"index":"timeId"})
index_item_pca_df.columns = ["timename"] + ["i"+str(i) for i in range(30)]
index_item_pca_df
```

```python
index_item_pca_df.to_csv('./timeVectors_{}.csv'.format(n_components), sep=' ', index=False, header=True)
```
