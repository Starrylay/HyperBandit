```python
import  numpy as np
import json
```

```python

FileName = "./user_friends.dat.mapped.dat"
dict ={}
with open(FileName, "r") as f:
    i = 0
    for line in f:
        i+=1
        # line = line.split("\t")
        num = int(line) #[0] 是item id [1]是arm feature
        dict[i] = num
FileName1 = "./cluster_map.json"
with open(FileName1, "w") as f:
    json.dump(dict,f)
    print("done")
```
